CREATE CATALOG

```python
spark.sql("""
CREATE CATALOG IF NOT EXISTS openbanking
MANAGED LOCATION
'abfss://databricks@graxianodatalake.dfs.core.windows.net/openbanking/'
COMMENT 'Lakehouse personal de cuentas y tarjeta bancaria'
""")
```

CREATE SCHEMAS

```python
for schema in ("bronze", "silver", "gold", "ops"):
    spark.sql(
        f"CREATE SCHEMA IF NOT EXISTS openbanking.{schema}"
    )

display(spark.sql("DESCRIBE CATALOG EXTENDED openbanking"))
display(spark.sql("SHOW SCHEMAS IN openbanking"))
```
